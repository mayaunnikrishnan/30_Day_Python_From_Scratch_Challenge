```Q1. Recreate the len() function
Question:
Write your own function called my_len() that counts the number of elements in an iterable without
using Python's built-in len() function.
Example:
my_len([10, 20, 30, 40, 50])Output: 5
Hint: Use a counter variable and a for loop.

````python
def my_len(item):
    count=0
    for i in item:
        count+=1
    print(count)
````

````python
my_len([10, 20, 30, 40, 50])
````

````python
my_len('my string')
````

```Q2. Multiplication Table
Question:
Take a number from the user and generate its multiplication table from 1 to 10.
Example:
Input: 5

5 x 1 = 5
5 x 2 = 10
5 x 3 = 15
...
5 x 10 = 50
Hint: Use a loop and a variable to keep track of the multiplier.


````python
def multiplication_table(number):
    for i in range(1,11):
        print(f"{number} * {i} = {i*number}")
````

````python
multiplication_table(5)
````

```Q3. Count the Number of Digits
Question:
Take an integer from the user and count the total number of digits without converting the number into a string.
Example:
Input: 58392
Output: 5
Hint: Repeatedly divide the number by 10 using integer division. Consider the special case of zero.


````python
def count_number(number):
    count=0
    if number==0:
        count=1
    else:
        while number!=0:
            number=abs(number//10)
            count+=1
    return(count)
number=int(input('Enter a number : '))
count_number(number)
````

```Q4. Factorial Using Iteration
Question:
Write a function to calculate the factorial of a given number using a loop.
Example:
Input: 5
Output: 120
Hint: Initialize the result to 1 and multiply it by each number from 1 to n.


````python
def factorial(number):
    if number ==0:
        return 1
    else:
        count=1
        for i in range(1,number+1):
            count*=i
        return count
number=int(input('Enter a number : '))
factorial(number)
````
