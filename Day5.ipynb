Q1. Recreate the any() function
Question:
Create your own function called my_any() that behaves like
Python's built-in any() function without using any().
Example:
my_any([0, 0, 5, 0])Output: Truemy_any([0, 0, 0])Output: False
Hint: Use a loop and return True immediately when you encounter a truthy element.


````python
def my_any(items):
    for item in items:
        if item:
            return True
    else:
        return False
my_any([0, 0, 5, 0])
````

````python
my_any([0, 0, 0])
````

Q2. Check Whether a List Contains an Even Number
Question:
Given a list of integers, check whether at least one even number exists.
Example:
numbers = [3, 7, 9, 12, 15]Output: True
Hint: Use the modulus operator (%) and stop when you find an even number.


````python
def any_even(items):
    for item in items:
        if item%2==0:
            return True
    else:
        return False
any_even([3, 7, 9, 12, 15])
````

````python
any_even([3, 7, 9, 15])
````

Q3. Check Whether a Number Contains a Zero
Question:
Take an integer from the user and check whether any of its digits is zero.
Example:
Input: 58302
Output: True

Input: 58321
Output: False
Hint: Extract digits using modulus and integer division.
Think about how to stop as soon as zero is found.


````python
def contain_zero(number):
    while number!=0:
        item=number%10
        if not item:
            return True
        number=number//10
    else:
        return False
contain_zero(58302)
````

````python
contain_zero(58321)
````

Q4. Check Whether a Sentence Contains a Vowel
Question:
Take a sentence from the user and check whether it contains at least one vowel (a, e, i, o, u).
Example:
Input: "Python programming"

Output: True
Hint: Iterate through the characters and check whether each character belongs to the set of vowels.


````python
def check_vowel(sentence):
    for i in sentence:
        if i.lower() in 'aeiou':
            return True
    else:
        return False
check_vowel("Python programming")
````

````python
print(check_vowel("rhythm"))
````
