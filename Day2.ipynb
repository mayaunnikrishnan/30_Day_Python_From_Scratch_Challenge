```Q1. Recreate the sum() function
Question:
Create your own function called my_sum() that calculates the sum of all elements
in a list without using the built-in sum() function.
Example:
my_sum([10, 20, 30, 40])Output: 100
Hint: Use an accumulator variable.


````python
def my_sum(my_list):
    total=0
    for i in my_list:
        total+=i
    return total
my_sum([10, 20, 30, 40])
````

```Q2. Sum of Natural Numbers
Question:
Take a positive integer n from the user and calculate the sum of all
natural numbers from 1 to n.
Example:
Input: 5
Output: 15

1 + 2 + 3 + 4 + 5 = 15
Hint: Start with total = 0 and add each number during iteration.


````python
def total_sum(number):
    total=0
    for i in range(1,number+1):
        total+=i
    return total
number=int(input('Enter the number : '))
total_sum(number)
````

```Q4. Fibonacci Series Using Iteration
Question:
Generate the first n terms of the Fibonacci series.
Example:
Input: 7

Output:
0 1 1 2 3 5 8
Hint: Maintain two variables representing the previous two numbers.
 Calculate the next number by adding them, then update both variables.


````python
def fibonacci(number):
    a,b=0,1
    for i in range(0,number):
        print(a,end=" ")
        a,b=b,a+b
number=int(input('Enter the number : '))
fibonacci(number)
````
