```Q1. Recreate the max() function
Question:
Create your own function called my_max() that finds the
largest element in a list without using the built-in max() function.
Example:
my_max([12, 45, 7, 89, 23])Output: 89
Hint: Initialize a variable with the first element and compare the remaining elements.


````python
def my_max(elements):
    max_number=elements[0]
    for num in elements:
        if num>max_number:
            max_number=num
    return max_number
my_max([12, 45, 7, 89, 23])
````

```Q2. Find the Largest Digit
Question:
Take an integer from the user and find the largest digit in that number.
Example:
Input: 58392
Output: 9
Hint: Extract individual digits using modulus (%) and integer division (//).


````python
def largest_digit(number):
    max_number=number%10
    while number!=0:

        if (number%10)>max_number:
            max_number=number%10
        number=number//10

    return max_number
largest_digit(58392)
````

```Q3. Find the Second Largest Distinct Number
Question:
Find the second largest distinct number in a list without using sort() or max().
Example:
numbers = [10, 25, 45, 45, 30, 20]Output: 30
Hint: Maintain two variables: largest and second_largest. Handle duplicate values carefully.


````python
def second_largest(elements):
    if not elements :
        return 'empty list'
    else:
        max_number=elements[0]
    second_number=float('-inf')
    for i in elements:
        if i>max_number:
            second_number=max_number
            max_number=i
        elif i<max_number and i>second_number:
            second_number=i
    if second_number==float('-inf'):
        return 'No second largest'
    return second_number
second_largest([10, 25, 45, 45, 30, 20])
````

```Q4. Find the Student with the Highest Marks
Question:
Given a dictionary containing student names and their marks,
find the student who scored the highest marks.
Example:
marks = {    "Anu": 85,    "Ravi": 92,    "Meera": 78,    "Arun": 96}Output:Arun: 96
Hint: Iterate through dictionary items and compare the marks.


````python
def highest_mark(kwargs):
    max_number=list(kwargs.values())[0]
    name=list(kwargs.keys())[0]
    for i,v in kwargs.items():
        if v>max_number:
            max_number=v
            name=i
    return name,max_number
highest_mark({    "Anu": 85,    "Ravi": 92,    "Meera": 78,    "Arun": 96})
````
