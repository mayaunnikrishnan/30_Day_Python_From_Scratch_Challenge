Q1. Recreate the abs() function
Question:
Create your own function called my_abs() without using Python's built-in abs() function.
Example:
my_abs(-25)Output: 25my_abs(15)Output: 15my_abs(0)Output: 0
Hint: Use an if-else condition.


````python
def my_abs(number):
    if number<0:
        return -number
    else:
        return number
````

````python
#using conditional expression
def my_abs(number):
    return -number if number<0 else number
````

````python
my_abs(-25)
````

````python
my_abs(15)
````

````python
my_abs(0)
````

Q2. Find the Absolute Difference
Question:
Take two numbers from the user and calculate their absolute difference without using abs().
Example:
Input:
a = 15
b = 40

Output: 25
Hint: Subtract the smaller number from the larger number.


````python
def abs_diff(a,b):
    if a>b:
        return a-b
    else:
        return b-a
a=int(input('enter a number : '))
b=int(input('enter a number : '))
abs_diff(a,b)
````

````python
#using conditional expression
def abs_diff(a,b):
    return a-b if a>b else b-a
````

Q3. Calculate the Distance Between Two Points
Question:
Given two points on a number line, calculate the distance between them.
Do not use abs().
Example:
Input:
Point A = -8
Point B = 5

Output: 13
Hint: Distance is the absolute difference between the two coordinates.


````python
my_abs=lambda x:-x if x<0 else x
def dis_btw_points(a,b):
    return my_abs(a-b)
a=int(input('enter a number : '))
b=int(input('enter a number : '))
dis_btw_points(a,b)
````

Q4. Find the Number Closest to Zero
Question:
Given a list containing positive and negative integers,
find the number closest to zero without using abs() or min().
Example:
numbers = [-15, 8, -3, 12, 5]Output: -3
Hint: Compare the magnitudes of numbers.
Handle positive and negative values separately.


````python
my_abs=lambda x: -x if x<0 else x
def closest_to_zero(elements):
    smallest= elements[0]
    for i in elements:
        if my_abs(i)<my_abs(smallest):
            smallest=i
    return smallest
closest_to_zero([-15, 8, -3, 12, 5])
````
