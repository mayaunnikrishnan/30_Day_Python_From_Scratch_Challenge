Q1. Recreate the all() function
Question:
Create your own function called my_all() that behaves like Python's built-in all() function without using all().
Example:
my_all([1, 2, 3, 4])Output: Truemy_all([1, 0, 3, 4])Output: Falsemy_all([])Output: True
Hint: Use a loop and return False immediately when you encounter a falsy element. Remember the behaviour of an empty iterable.



````python
def my_all(elements):
    for i in elements:
        if not i:
            return False
    else:
        return True
my_all([1, 2, 3, 4])
````

````python
my_all([1, 0, 3, 4])
````

````python
my_all([])
````

Q2. Check Whether All Numbers Are Positive
Question:
Given a list of numbers, check whether all numbers are greater than zero.
Example:
numbers = [10, 20, 30, 40]Output: True
Hint: Use a condition inside a loop. Stop when you find a number that is not positive.


````python
def positive_number(elements):
    for i in elements:
        if i <= 0:
            return False
    else:
        return True
positive_number([10, 20, 30, 40])
````

````python
positive_number([10, -20, 30, 40])
````

Q3. Check Whether a String Contains Only Alphabets
Question:
Take a string from the user and check whether every
character is an alphabet.
Do not use the built-in isalpha() method.
Example:
Input: "Python"
Output: True

Input: "Python123"
Output: False
Hint: Compare each character with the ranges of
uppercase and lowercase ASCII letters using ord().


````python
def check_string(ele):
    for i in ele:
        if i.lower() not in 'abcdefghijklmnopqrstuvwxyz':
            return False
    else:
        return True
check_string("Python")
````

````python
check_string("Python123")
````

````python
#Another method
def check_string(ele):
    for i in ele:
        if not(65<=ord(i)<= 90 or 97<=ord(i)<=122):
            return False
    else:
        return True
check_string("Python")
````

Q4. Validate Student Marks
Question:
You are given a list of students' marks.
Check whether every student has passed the examination.
Passing mark: 40.
Example:
marks = [65, 78, 45, 82, 55]Output: True

marks = [65, 32, 45, 82, 55]Output: False
Hint: Combine iteration and conditional checking.
Return False if any mark is below 40.


````python
def ispassed(elements):
    for i in elements:
        if i <40:
            return False
    else:
        return True
ispassed([65, 78, 45, 82, 55])
````

````python
ispassed([65, 32, 45, 82, 55])
````
