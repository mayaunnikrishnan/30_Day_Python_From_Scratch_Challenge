## Q1. Recreate the min() function

**Question:**

Create your own function called `my_min()` that finds the smallest element in a list without using the built-in `min()` function.

**Example:**

```python
my_min([45, 12, 78, 5, 23])
```

**Output:**
```text
5
```

**Hint:** Use the same comparison technique from Day 3, but reverse the comparison condition.


````python
def my_min(elements):
    min_val=elements[0]
    for i in elements:
        if i<min_val:
            min_val=i
    return min_val
my_min([45, 12, 78, 5, 23])
````


## Q2. Find the Smallest Digit

**Question:**

Take an integer from the user and find the smallest digit in that number.

**Example:**

```text
Input: 58392
```

**Output:**
```text
2
```

**Hint:** Extract each digit and compare it with the current smallest digit.


````python
def smallest_integer(number):
    min_val=number%10
    while number!=0:
        if number%10<min_val:
            min_val=number%10
        number=number//10
    return min_val
smallest_integer(58392)
````


## Q3. Find the Second Smallest Distinct Number

**Question:**

Find the second smallest distinct number in a list without using `sort()` or `min()`.

**Example:**

```python
numbers = [15, 5, 20, 5, 10, 25]
```

**Output:**
```text
10
```

**Hint:** Maintain two variables: `smallest` and `second_smallest`.


````python
def second_smallest(elements):
    min_val=elements[0]
    second_min=float('inf')
    for i in elements:
        if i<min_val:
            second_min=min_val
            min_val=i
        elif i>min_val and i<second_min:
            second_min=i
    return second_min
second_smallest([15, 5, 20, 5, 10, 25])
````


## Q4. Find the Closest Number to Zero

**Question:**

Given a list containing positive and negative numbers, find the number closest to zero without using `abs()` or `min()`.

**Example:**

```python
numbers = [-8, 4, -2, 10, 3]
```

**Output:**
```text
-2
```

**Hint:** Compare the magnitude of each number. You can calculate magnitude manually using a conditional expression.


````python
def closest_tozero(elements):
    magnitude=lambda x:-x if x<0 else x
    min_val=elements[0]
    for i in elements:
        if magnitude(i)<magnitude(min_val):
            min_val=i
    return min_val
closest_tozero([-8, 4, -2, 10, 3])
````
